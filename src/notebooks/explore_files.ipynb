## Les imports

```python
import os
from pathlib import Path
from dotenv import load_dotenv
from collections import Counter

load_dotenv()
```

## Les fichiers 

```python
raw_dir = Path("../..") / os.environ["GARMIN_FIT_DIR"]

print("Répertoire :", raw_dir.resolve())

files = [
    file for file in raw_dir.rglob("*")
    if file.is_file()
]

print(f"Nombre total de fichiers : {len(files)}")

extensions = Counter(file.suffix.lower() for file in files)
print(extensions)

for extension, count in extensions.most_common():
    print(f"{extension or '[sans extension]'} : {count}")
```
